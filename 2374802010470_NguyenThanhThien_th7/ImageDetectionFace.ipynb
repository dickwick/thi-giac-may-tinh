```python
import cv2
import matplotlib.pyplot as Thien

img = cv2.imread('hinh1.jpg')

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

Thien.imshow(img_rgb)
Thien.title("Original Image")
Thien.axis('off')
```

```python
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

faces = face_cascade.detectMultiScale(img_gray, 1.3, 5)

img_face = img_rgb.copy()

for (x, y, w, h) in faces:
    cv2.rectangle(img_face, (x,y), (x+w,y+h), (255,0,0), 3)

Thien.imshow(img_face)
Thien.title("Face Detection")
Thien.axis('off')
```

```python
eye_cascade = cv2.CascadeClassifier('haarcascade_eye.xml')

eyes = eye_cascade.detectMultiScale(img_gray, 1.3, 5)

img_eye = img_rgb.copy()

for (x, y, w, h) in eyes:
    cv2.rectangle(img_eye, (x,y), (x+w,y+h), (0,255,0), 2)

Thien.imshow(img_eye)
Thien.title("Eye Detection")
Thien.axis('off')
```

```python
smile_cascade = cv2.CascadeClassifier('haarcascade_smile.xml')

smiles = smile_cascade.detectMultiScale(img_gray, 1.7, 20)

img_smile = img_rgb.copy()

for (x, y, w, h) in smiles:
    cv2.rectangle(img_smile, (x,y), (x+w,y+h), (255,255,0), 2)

Thien.imshow(img_smile)
Thien.title("Smile Detection")
Thien.axis('off')
```
